```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input, Concatenate
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.preprocessing import RobustScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import tensorflow as tf
import random

# 设置图表样式
sns.set_style("whitegrid")
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ===== 函数定义 =====
def load_and_merge_data(filepath):
    df = pd.read_csv(filepath)
    merged_df = df.groupby(['year', 'country'])['count'].sum().reset_index()
    return merged_df

def plot_yearly_countries(df):
    yearly_counts = df.groupby('year')['country'].nunique().reset_index()
    plt.figure(figsize=(14, 7))
    ax = sns.lineplot(data=yearly_counts, x='year', y='country', marker='o', markersize=8, linewidth=2.5, color='#2c7bb6')
    plt.title('Number of Unique Countries Appearing Each Year', fontsize=16, pad=20)
    plt.xlabel('Year', fontsize=14)
    plt.ylabel('Number of Countries', fontsize=14)
    plt.xticks(range(1992, 2021, 2), rotation=45)
    for x, y in zip(yearly_counts['year'], yearly_counts['country']):
        ax.text(x, y+0.5, f'{y}', ha='center', va='bottom', fontsize=10)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
    return yearly_counts

# =========================
# 改进版训练和预测函数（取消差分）
def train_and_predict(data, n_future=10, return_eval=True):
    # 不进行差分，直接用原始序列进行建模
    if isinstance(data, (pd.DataFrame, pd.Series)):
        series = np.array(data).flatten().astype(float)
    else:
        series = np.array(data).flatten().astype(float)

    scaler = RobustScaler()
    scaled_data = scaler.fit_transform(series.reshape(-1,1))

    n_steps = 8
    X, y, X_feat = [], [], []

    # 构造序列
    for i in range(n_steps, len(scaled_data)):
        X.append(scaled_data[i - n_steps:i, 0])  # 时间序列部分
        y.append(scaled_data[i, 0])               # 目标
        # 简单示例：加入年份信息作为特征（可扩展为其他特征）
        year_feature = np.array([[i]])  # 可以自定义，示例用索引作为特征
        X_feat.append(year_feature.flatten())

    X, y, X_feat = np.array(X), np.array(y), np.array(X_feat)

    # 归一化特征（如果特征多于一维，可以用Scaler）
    # (这里暂时不单独归一化，保持简单)

    # 训练/测试划分
    split_idx = int(len(X) * 0.8)
    X_train, X_test = X[:split_idx], X[split_idx:]
    X_feat_train, X_feat_test = X_feat[:split_idx], X_feat[split_idx:]
    y_train, y_test = y[:split_idx], y[split_idx:]

    # 构建模型
    seq_input = Input(shape=(n_steps,1), name='sequence_input')
    feat_input = Input(shape=(X_feat.shape[1],), name='features_input')

    x = LSTM(128, return_sequences=True)(seq_input)
    x = Dropout(0.2)(x)
    x = LSTM(64)(x)

    combined = Concatenate()([x, feat_input])
    x2 = Dense(64, activation='relu')(combined)
    x2 = Dropout(0.2)(x2)
    output = Dense(1)(x2)

    model = Model(inputs=[seq_input, feat_input], outputs=output)
    model.compile(optimizer='adam', loss='mse')
    early_stop = EarlyStopping(monitor='val_loss', patience=30, restore_best_weights=True)

    model.fit([X_train.reshape(-1, n_steps,1), X_feat_train], y_train,
              epochs=400, batch_size=8,
              validation_data=([X_test.reshape(-1, n_steps,1), X_feat_test], y_test),
              callbacks=[early_stop], verbose=0)

    # 评估指标（训练集）
    y_train_pred = model.predict([X_train.reshape(-1, n_steps,1), X_feat_train], verbose=0).flatten()
    y_train_pred_inv = scaler.inverse_transform(y_train_pred.reshape(-1,1)).flatten()
    y_train_true_inv = scaler.inverse_transform(y_train.reshape(-1,1)).flatten()

    train_eval = {
        'mse': mean_squared_error(y_train_true_inv, y_train_pred_inv),
        'mae': mean_absolute_error(y_train_true_inv, y_train_pred_inv),
        'r2' : r2_score(y_train_true_inv, y_train_pred_inv)
    }

    # 测试集
    y_test_pred = model.predict([X_test.reshape(-1, n_steps,1), X_feat_test], verbose=0).flatten()
    y_test_pred_inv = scaler.inverse_transform(y_test_pred.reshape(-1,1)).flatten()
    y_test_true_inv = scaler.inverse_transform(y_test.reshape(-1,1)).flatten()

    test_eval = {
        'mse': mean_squared_error(y_test_true_inv, y_test_pred_inv),
        'mae': mean_absolute_error(y_test_true_inv, y_test_pred_inv),
        'r2' : r2_score(y_test_true_inv, y_test_pred_inv)
    }

    # 未来预测
    last_seq = scaled_data[-n_steps:]
    future_preds_scaled = []

    for _ in range(n_future):
        x_input = last_seq.reshape(1, n_steps, 1)
        # 未来特征（这里用简单的索引，可以自己定义）
        year_fea = np.array([[len(scaled_data)+_]])
        pred_scaled = model.predict([x_input, year_fea.flatten().reshape(1, -1)], verbose=0)[0,0]
        future_preds_scaled.append(pred_scaled)
        last_seq = np.append(last_seq[1:], pred_scaled)

    predictions = scaler.inverse_transform(np.array(future_preds_scaled).reshape(-1,1)).flatten()

    if return_eval:
        return predictions, train_eval, test_eval
    else:
        return predictions

def plot_country_predictions(actual, predictions, country):

    actual = np.array(actual).flatten().astype(float)
    predictions = np.array(predictions).flatten().astype(float)

    years = np.arange(1992, 2021)
    future_years = np.arange(2021, 2021 + len(predictions))

    plt.figure(figsize=(12, 6))

    # 🔵 Historical
    plt.plot(years, actual, 'bo-', 
             label=f"{country} – Historical Appearances")

    # 标注历史
    for x, y in zip(years, actual):
        plt.text(x, y + 0.3, str(int(round(y))),
                 ha='center', va='bottom', fontsize=8)

    # 🔴 Predicted
    plt.plot(future_years, predictions, 'ro-', 
             label=f"{country} – Predicted Appearances")

    # 标注预测
    for x, y in zip(future_years, predictions):
        plt.text(x, y + 0.3, str(int(round(y))),
                 ha='center', va='bottom', fontsize=8)

    plt.xlabel("Year")
    plt.ylabel("Appearance Count")

    # ⭐ 标题
    plt.title(f"{country}: Historical vs Predicted Appearances")

    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend()
    plt.tight_layout()
    plt.show()
def get_top5_countries(df):
    top5_countries = df.groupby('country')['count'].sum().nlargest(5).index.tolist()
    return top5_countries

def prepare_country_data(df, country):
    country_data = df[df['country'] == country].groupby('year')['count'].sum().reset_index()
    full_years = pd.DataFrame({'year': range(1992, 2021)})
    country_data = pd.merge(full_years, country_data, on='year', how='left').fillna(0)
    return country_data[['count']].values.astype(float).flatten()

def plot_unique_countries(actual, predictions):

    actual = np.array(actual).flatten().astype(float)
    predictions = np.array(predictions).flatten().astype(float)

    years = np.arange(1992, 2021)
    future_years = np.arange(2021, 2021 + len(predictions))

    plt.figure(figsize=(12, 6))

    # Actual
    plt.plot(years, actual, 'bo-', label='Unique Countries Actual')

    # 👉 标注历史数据
    for x, y in zip(years, actual):
        plt.text(x, y + 0.3, str(int(round(y))),
                 ha='center', va='bottom', fontsize=8)

    # Predicted
    plt.plot(future_years, predictions, 'ro-', label='Unique Countries Predicted')

    # 👉 标注预测数据
    for x, y in zip(future_years, predictions):
        plt.text(x, y + 0.3, str(int(round(y))),
                 ha='center', va='bottom', fontsize=8)

    plt.xlabel("Year")
    plt.ylabel("Number of Countries")
    plt.title("Number of Unique Countries Appearing Each Year")

    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend()
    plt.tight_layout()
    plt.show()

# ===== 主函数 =====
if __name__ == "__main__":
    data_path = "number-of-nations.csv"
    merged_df = load_and_merge_data(data_path)

    print("\n🧠 Predicting future unique country counts...")
    country_count_predictions, train_eval, test_eval = train_and_predict(yearly_counts['country'])

    print("\n📊 Training Set Evaluation:")
    print(f"MSE: {train_eval['mse']:.3f}, MAE: {train_eval['mae']:.3f}, R²: {train_eval['r2']:.3f}")
    print("📊 Testing Set Evaluation:")
    print(f"MSE: {test_eval['mse']:.3f}, MAE: {test_eval['mae']:.3f}, R²: {test_eval['r2']:.3f}")

    print("\n📈 Predicted number of unique countries (2021-2030):")
    for year, val in zip(range(2021, 2031), country_count_predictions):
        print(f"{year}: {int(round(val))}")

    plot_unique_countries(
        yearly_counts['country'].values,
        country_count_predictions)

    # ===== Top5 国家分析与预测 =====
    top5_countries = get_top5_countries(merged_df)
    print("\n🏆 Top5 Countries by Total Appearances:")
    top5_totals = merged_df.groupby('country')['count'].sum().loc[top5_countries]
    for rank, (country, total) in enumerate(top5_totals.items(), 1):
        print(f"{rank}. {country} - Total Appearances: {int(total)}")
    
    # 分别预测每个 Top5 国家
    for country in top5_countries:
        print(f"\n🧠 Processing country: {country}")
        country_data = prepare_country_data(merged_df, country)
        predictions, train_eval, test_eval = train_and_predict(
            country_data, n_future=10)

        print(f"[{country}] Training  - MSE: {train_eval['mse']:.3f}, MAE: {train_eval['mae']:.3f}, R²: {train_eval['r2']:.3f}")
        print(f"[{country}] Testing   - MSE: {test_eval['mse']:.3f}, MAE: {test_eval['mae']:.3f}, R²: {test_eval['r2']:.3f}")

        # 未来十年预测数据单独列出
        print(f"📈 {country} Predicted Appearances (2021-2030):")
        for year, val in zip(range(2021, 2031), predictions):
            print(f"{year}: {int(round(val))}")

        # 绘制历史数据和预测曲线
        plot_country_predictions(country_data, predictions, country)
```
